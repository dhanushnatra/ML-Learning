## Partial Derivatives 

eqn Z = X**2 - y **2

D'Z/D'X = 2x ( by taking y as contant since we are differentiating w.r to x)

D'Z/D'y = -2y ( by taking x as contant since we are differentiating w.r to y)


```python
import numpy as np
import matplotlib.pyplot as plt


x = np.linspace(-3,3,100)
def f(my_x,my_y):
    return my_x**2  - my_y**2 

zs_wr_x=f(x,0)

def derv_x(my_x):
    return 2*my_x 

def derv_y(my_y):
    return -2*my_y 

def plot_tangent_wrt_x(my_xs,my_x,my_y,color):
    
    tan_aux_y = f(my_xs,my_y)
    plt.scatter(my_xs,tan_aux_y,c=color,zorder=3)
    
    tan_m=derv_x(my_xs)
    tan_b = tan_aux_y-tan_m*my_xs
    tan_y = tan_m*my_x + tan_b
    plt.plot(x,tan_y,c=color,zorder=3,linestyle="--",linewidth=0.9)
```

```python
x_samples = [-2,-1,0,1,2]
x_colors = ["red","orange","green","blue","lightblue"]

fig,ax = plt.subplots()
plt.ylim(-1,9)

for i,x_sam in enumerate(x_samples):
    plot_tangent_wrt_x(x_sam,x,0,x_colors[i])
    
plt.plot(x,zs_wr_x)
```

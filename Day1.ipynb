```python
top_popular_movies = movies.sort_values(by="popularity",ascending=False).head(10)
top_rated_movies=movies.sort_values(by="rating",ascending=False).head(10)
top_voted_movies=movies.sort_values(by="vote_count",ascending=False).head(10)


top_popular_tv_shows = tv_shows.sort_values(by="popularity",ascending=False).head(10)
top_rated_tv_shows=tv_shows.sort_values(by="rating",ascending=False).head(10)
top_voted_shows=tv_shows.sort_values(by="vote_count",ascending=False).head(10)
```

```python
def annotate(texts,x,y,ax):
    for i,text in enumerate(texts):
        ax.annotate(
            text=text,
            xy=(x[i],y[i]),
            fontsize=8
        )
def barh(x,y,title,ax,line_color="r",xlabel=None,ylabel=None):
    ax.barh(x,y,color=line_color)
    ax.set_title(title)
    if xlabel!=None:
        ax.set_xlabel(xlabel)
    if ylabel!=None:
        ax.set_ylabel(ylabel)

def plot(x,y,title,ax,linewidth=1.5,line_color="r",xlabel=None,ylabel=None):
    ax.plot(x,y,linewidth=linewidth,color=line_color)
    ax.set_title(title)
    if xlabel!=None:
        
        ax.set_xlabel(xlabel)
    if ylabel!=None:
        ax.set_ylabel(ylabel)
    
    
```

```python
import warnings

warnings.filterwarnings("ignore")

fig,ax=plt.subplots(3,2,figsize=(30,20))

barh(
    top_rated_movies["title"],
    top_rated_movies["rating"],
    "top Rated movies",ax[0,0]
)


barh(
    top_popular_movies["title"],
    top_popular_movies["popularity"],
    "popular Movies",ax[1,0]
)

barh(
    top_voted_movies["title"],
    top_voted_movies["vote_count"],
    "top Voted Movies",ax[2,0]
)

barh(
    top_rated_tv_shows["title"],
    top_rated_tv_shows["rating"],
    "top rated tv shows",ax[0,1]
)

barh(
    top_popular_tv_shows["title"],
    top_popular_tv_shows["popularity"],
    "popular tv shows",ax[1,1]
)

barh(
    top_voted_shows["title"],
    top_voted_shows["vote_count"],
    "top voted shows",ax[2,1]
)
```

### what is probability
probability p(event) = num of outcomes of a event / num of total outcomes

example:

probability of getting a head in coin flip = 1/2 


double coin flip

P(HH) = probability of getting 2 heads consecutively = 1/4 = 0.25 = (1/2)*(1/2)

P(HHH) = (1/2)*(1/2) *(1/2) = 1/8 = 0.125

### Combinatronics

4! = 4x3x2x1 = 24

number of outcomes of a event = (n  k) = n! / k!(n-k)!


(3  2)= 3!/2!(3-2)!  = 3x2x1 / (2x1)(1) = 6/2 = 3


p(event) = 3/ 2^n = 3/2^3 = 0.375

```python
from math import factorial

def coin_flip_prob(n,k):
    n_choose_k = factorial(n)/(factorial(k)*factorial(n-k))
    return n_choose_k/2**n

coin_flip_prob(5,3)
```

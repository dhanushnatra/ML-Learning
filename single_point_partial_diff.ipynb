```python
import torch
import matplotlib.pyplot as plt
```

```python



xs=torch.tensor([0,1,2,3,4,5,6,7.])
ys = torch.tensor([1.86, 1.31, .62, .33, .09, -.67, -1.23, -1.37])
i=7
x=xs[i]
y=ys[i]
```

```python
def regression(my_m,my_x,my_b):
    return (my_m*my_x)+my_b

m = torch.tensor([0.9]).requires_grad_()
b = torch.tensor([0.1]).requires_grad_()
m,b
```

```python
y_hat = regression(m,x,b)
y_hat
```

```python
def mse_single(my_yhat,my_y):
    return (my_yhat-my_y)**2

C = mse_single(y_hat,y)
C.backward()
m.grad,b.grad
```

### Gradient (nabla C )

list of all the del C / del model parameters with transpose applied to make them in a single column


nabla C = [del C / del p1 ,del C / del p2 , del C /del p3 ........del C / del pm].T 

```python
gradient = torch.tensor([[b.grad.item(),m.grad.item()]]).T
gradient
```

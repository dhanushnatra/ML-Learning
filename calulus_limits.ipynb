lim(x->1) x\*\*2-1 / x-1


```python
def lim_func(my_x):
    my_y= (my_x**2-1)/(my_x-1)
    return my_y


lim_func(2)
```

```python
y=lim_func(x)


fig,ax=plt.subplots()
plt.axvline(x=1,color="purple",linestyle="--")
plt.axhline(y=2,color="purple",linestyle="--")
ax.set_xlim([-2.5,2.5])
ax.set_ylim([-2.5,4])
_=ax.plot(x,y)
```

lim(x->1) sinx/x


```python
def sin_lim(my_x):
    my_y = np.sin(my_x)/my_x
    return my_y

y=sin_lim(x)

fig,ax=plt.subplots(1,2,figsize=(10,6))

ax[1].axvline(x=0,color="purple",linestyle="--")
ax[1].axhline(y=1,color="purple",linestyle="--")
ax[1].set_xlim([-1,1])
ax[1].set_ylim([0.5,1.5])
ax[1].plot(x,y)
ax[0].plot(x,y)
```

lim(x->infinity) 25/x


```python
def infi_func(my_x):
     return 25/my_x
 
y=infi_func(x)


fig,ax=plt.subplots()

ax.axvline(x=0,color="lightgrey")

ax.axhline(y=0,color="lightgrey")
ax.set_xlim([-10,10])
ax.set_ylim([-300,300])
_=ax.plot(x,y)
```

```python
left_x=x[x<0]
right_x=x[x>0]

left_y=infi_func(left_x)
right_y=infi_func(right_x)


fig,ax=plt.subplots()

ax.axvline(x=0,color="lightgrey")
ax.axhline(y=0,color="lightgrey")

ax.set_xlim([-10,10])
ax.set_ylim([-300,300])
ax.plot(left_x,left_y,color="C0")
ax.plot(right_x,right_y,color="C0")
```

2. lim(x->4) x\*\*2-2x-8 / x-4


```python
def lim_1(my_x):
    return (my_x**2-25)/(my_x+5)
def lim_2(my_x):
    return (my_x**2-2*my_x-8)/(my_x-4)
```

## slope of a point in a parabola

### Steps:

1. make a linspace distribution values
2. create a parabola ( polynomial - bionomial distribution )
3. plot the parabola
4. find the nearest point from the given point ( delta_x ~= 0)
5. find m
6. find the b ( from y=mx+b => b=y-mx )
7. find the y by using the linspace distribution values


#### x\**2 + 2*x + 2


```python
def f(x_):
    return x_**2 + 2*x_ + 2

y=f(x)

fig,ax = plt.subplots()

_=ax.plot(x,y)
```

```python
x1 = -1

delta_x=0.0000001

x2 = x1+delta_x

y2 = f(x2)
y1=f(x1)

m = (y2 - y1 )/ delta_x

b =y1- m*x1 

y_tan_line = m*x + b 

fig,ax = plt.subplots()
plt.scatter(x=x1,y=y1,c="orange",zorder=3)
plt.plot(x,y_tan_line,c="orange",zorder=3)
ax.set_xlim([-12,10])
ax.set_ylim([-20,130])
_=ax.plot(x,y)
```
